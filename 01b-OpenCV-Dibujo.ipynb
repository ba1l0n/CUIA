# Dibujo

OpenCV ofrece funciones básicas de dibujado que principalmente serán usadas en prototipos o con fines de depurado. Si queremos mayor calidad y versatilidad en los dibujos es mejor usar una biblioteca especializada (por ejemplo [Pillow](https://pypi.org/project/pillow/)).

```python
import cv2
import numpy as np
import cuia
```

## Dibujo de marcadores

[**drawMarker**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga644c4a170d4799a56b29f864ce984b7e)(Imagen, Posición, Color, Tipo, Tamaño, Grosor)
* Imagen     (Imagen  sobre la que se va a dibujar)
* Posición   ((x,y) ej. (200,300) )  
* Color      (ej. (0, 255,0) que representa el color verde en formato BGR)
* Tipo       (Tipo de [marcador](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga0ad87faebef1039ec957737ecc633b7b) )
* Tamaño     (Tamaño, ej 4. 
* Grosor     (1 por defecto))

```python
mandril = cv2.imread("media/baboon.jpg")
```

```python
imagen = mandril.copy()
cv2.drawMarker(imagen, (250,300), (255, 0, 0), cv2.MARKER_STAR, 20, 2)
cuia.plot(imagen)
```

## Dibujo de lineas

[**line**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga7078a9fae8c7e7d13d24dac2520ae4a2)(imagen, Inicio, Fin, Color, Grosor, Tipo, Shift)
* Imagen     (Imagen  sobre la que se va a dibujar)
* Inicio  ((x,y) ej. (100,100) )  
* Fin     ((x,y) ej. (200,300) )  
* Color   (ej. (255,0,0) que representa el color azul en formato BGR)  
* Grosor  (1 por defecto)  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Shift   (Fracción de pixel de desplazamiento de las coordenadas de los extremos de la linea, por defecto 0)

```python
imagen = mandril.copy()
cv2.line(imagen, (100,100), (200,300), (255,0,0), 4, cv2.LINE_4)
cuia.plot(imagen)
```

## Dibujo de rectángulos

Estos rectángulos están siempre alineados con los ejes

[**rectangle**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga07d2f74cadcf8e305e810ce8eed13bc9)(Imagen, Esquina1, Esquina2, Color, Grosor, Tipo, Shift)
* Imagen     (Imagen  sobre la que se va a dibujar)
* Esquina1(ej. (250,200) )  
* Esquina2(ej. (450,300) esquina opuesta)  
* Color   (ej. (0,255,0) que representa el color verde en formato BGR)  
* Grosor  (1 por defecto, si el valor es -1 se rellena el rectángulo)  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Shift   (Fracción de pixel de desplazamiento de las coordenadas de las esquinas, por defecto 0)

```python
imagen = mandril.copy()
cv2.rectangle(imagen, (250,200), (450,300), (0,255,0), -1, cv2.LINE_4, 0)
cuia.plot(imagen)
```

## Dibujo de círculos y circunferencias

[**circle**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf10604b069374903dbd0f0488cb43670)(Imagen, Centro, Radio, Color, Grosor, Tipo, Shift)
* Imagen     (Imagen  sobre la que se va a dibujar)
* Centro  (ej. (256,256) )  
* Radio   (ej. 150)  
* Color   (ej. (0,0,255) que representa el color rojo en formato BGR)  
* Grosor  (1 por defecto, si el valor es -1 se rellena el círculo)  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Shift   (Fracción de pixel de desplazamiento de las coordenadas del centro y el radio, por defecto 0)

```python
imagen = mandril.copy()
cv2.circle(imagen, (256,256), 150, (0,128,255), 5)
cuia.plot(imagen)
```

## Dibujo de elipses

[**ellipse**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga28b2267d35786f5f890ca167236cbc69)(Imagen, Centro, Radios, Ángulo, Inicio, Fin, Color, Grosor, Tipo, Shift)
* Imagen     (Imagen  sobre la que se va a dibujar)
* Centro  ((x,y) ej. (256,256) )  
* Radios  ((rx, ry) ej. (100,30))  
* Ángulo  (Ángulo de rotación de la elipse en sentido horario, ej. 45)  
* Inicio  (Ángulo de inicio del arco, empezando en el "norte" y en sentido antihorario, ej. 0)  
* Fin     (Ángulo final del arco, ej 270)  
* Color   (ej. (0,0,255) que representa el color rojo en formato BGR)  
* Grosor  (1 por defecto, si el valor es -1 se rellena la elipse)  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Shift   (Fracción de pixel de desplazamiento de las coordenadas del centro y el radio, por defecto 0)

```python
imagen = mandril.copy()
cv2.ellipse(imagen, (256,256), (200,80), 45, 0, 270, (255,0,255), -1)
cuia.plot(imagen)
```

## Dibujo de polilíneas

[**polylines**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga1ea127ffbbb7e0bfc4fd6fd2eb64263c)(Imagen, Vértices, Cerrado, Color, Grosor, Tipo, Shift)
* Imagen     (Imagen  sobre la que se va a dibujar)
* Vértices(Array de vértices de la polilínea, ej [np.array([[100,100], [150,250], [300,400]])]
* Cerrado (Valor lógico que indica si debe conectar el primer y último punto)  
* Color   (Color en formato BGR)  
* Grosor  (1 por defecto, un valor negativo no rellena la polilínea. Para ello está la función [**fillPoly**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga311160e71d37e3b795324d097cb3a7dc))  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Shift   (Fracción de pixel de desplazamiento de las coordenadas del centro y el radio, por defecto 0)

```python
imagen = mandril.copy()
cv2.polylines(imagen, [np.array([[100,100], [150,250], [300,400]])], True, (0,255,255), 4)
cuia.plot(imagen)
```

## Dibujo de polilíneas rellenas

[**fillPoly**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga311160e71d37e3b795324d097cb3a7dc)(Imagen, Vértices, Color, Tipo, Shift, Offset)
* Imagen     (Imagen  sobre la que se va a dibujar)
* Vértices(Array de vértices de la polilínea, ej [np.array([[100,100], [150,250], [300,400]])] 
* Color   (Color en formato BGR)  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Shift   (Fracción de pixel de desplazamiento de las coordenadas del centro y el radio, por defecto 0)
* Offset  (Desplazamiento común a todos los vértices, por defecto 0)

```python
imagen = mandril.copy()
cv2.fillPoly(imagen, [np.array([[100,100], [150,250], [300,400]])], (0,255,255))
cuia.plot(imagen)
```

## Escritura de texto [Hershey](https://en.wikipedia.org/wiki/Hershey_fonts)

[**putText**](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga5126f47f883d730f633d74f07456c576)(Imagen, Texto, Origen, Tipo, Tamaño, Color, Grosor, Tipo, Invertir)
* Imagen (Imagen  sobre la que se va a dibujar)
* Texto  (Texto a escribir)
* Origen (vértice superior izquierdo del comienzo del texto, ej (256,100) )  
* Tipo de letra (fuentes [Hershey](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#ga0f9314ea6e35f99bb23f29567fc16e11), ej cv2.FONT_HERSHEY_SIMPLEX)  
* Tamaño de letra (ej. 2.5)  
* Color   (Color en formato BGR)  
* Grosor  (Grosor del trazo, 1 por defecto)  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Invertir(Valor lógico que indica si el origen está en la esquina inferior izquierda, por defecto False)  

```python
imagen = mandril.copy()
cv2.putText(imagen, "Mandril", (120,256), cv2.FONT_HERSHEY_SIMPLEX, 2.5, (255,255,255), 4, cv2.LINE_AA)
cuia.plot(imagen)
```

## Escritura de texto (True Type Fonts)

Por defecto la instalación de OpenCV no incluye soporte para fuentes TrueType por lo que será necesario realizar una compilación personalizada usando WITH_FREETYPE=ON para poder emplear las funcionalidades de módulo [feetype](https://docs.opencv.org/4.x/d9/dfa/classcv_1_1freetype_1_1FreeType2.html)

[**putText**](https://docs.opencv.org/4.x/d9/dfa/classcv_1_1freetype_1_1FreeType2.html#aba641f774c47a70eaeb76bf7aa865915)(Imagen, Texto, Origen, Tamaño, Color, Grosor, Tipo, Invertir)
* Imagen (Imagen  sobre la que se va a dibujar)
* Texto  (Texto a escribir)
* Origen (Vértice superior izquierdo del comienzo del texto, ej (256,100) )  
* Tamaño de letra (ej. 2.5)  
* Color   (Color en formato BGR)  
* Grosor  (Grosor del trazo, 1 por defecto)  
* Tipo    (Tipo de [linea](https://docs.opencv.org/4.x/d6/d6e/group__imgproc__draw.html#gaf076ef45de481ac96e0ab3dc2c29a777), por defecto cv2.LINE_8)  
* Invertir(Valor lógico que indica si el origen está en la esquina inferior izquierda, por defecto False)  

```python
try:
    import cv2.freetype as ft
    imagen = mandril.copy()
    ft = cv2.freetype.createFreeType2() #Creamos el contenedor de la fuente TTF
    ft.loadFontData(fontFileName='media/Alkatra-VariableFont_wght.ttf', idx=0) #Cargamos la fuente indicando el índice del subtipo
    ft.putText(imagen, 'Mandril', (120, 256), 60, (255,  255, 255), -1, cv2.LINE_AA, True)
    cuia.plot(imagen)
except:
    print("Es necesario compilar opencv con soporte para FreeType")
```

Otra opción para emplear fuentes TTF es usar el módulo [*Pillow*](https://pillow.readthedocs.io/en/stable/index.html) (*Python Imaging Library*)
>    pip install Pillow
> 
Para poder usar Pillow en OpenCV es necesareio hacer una conversión entre los formatos en que ambas almacenan las imágenes y tener en cuenta que Pillow representa el color usando RGB.

```python
from PIL import Image, ImageFont, ImageDraw
```

```python
imagen = mandril.copy()
imagenpil = Image.fromarray(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))  #Transformamos la imagen para ser usada con Pillow
fuente = ImageFont.truetype("media/Alkatra-VariableFont_wght.ttf", 50) #Cargamos la fuente TTF
draw = ImageDraw.Draw(imagenpil)  #Preparamos el lienzo para dibujar sobre la imagen de fondo
draw.text((170, 300), "Mandril", font=fuente, fill=(0,0,255), stroke_width=2, stroke_fill=(255,255,255))
imagen = cv2.cvtColor(np.asarray(imagenpil), cv2.COLOR_RGB2BGR) #Hacemos la transformación inversa de vuelta al formato de OpenCV
cuia.plot(imagen)
```
