## Instalación de Python en Windows

**1. Instalación del gestor de instalaciones de Python**

El ***Python install manager*** es la herramienta recomendada para instalar y gestionar las instalaciones de Python. Se puede descargar de la [Microsoft App Store](https://apps.microsoft.com/detail/9nq7512cxl7t?hl=es-ES&gl=ES) o desde la página de [Python Releases for Windows](https://www.python.org/downloads/windows/).

Tras la instalación dispondremos de un nuevo comando **[py](https://docs.python.org/es/3/using/windows.html)** en la terminal que, además de invocar a Python nos permitirá instalar y gestionar múltiples versiones de Python.

**2. Instalación de una versión específica de Python**

La herramienta **py** nos permite instalar y gestionar cualquier versión de Python desde la 3.5 en adelante. Por ejemplo, para instalar la versión 3.12 (recomendada en estas prácticas por cuestiones de compatibilidad con algunas bibliotecas):

```bash
py install 3.12
```

En todo momento podemos ver las versiones de Python que tenemos instaladas mediante **py list** o desinstalar alguna versión específica usando **py uninstall**

**3. Ejecución de una versión específica de Python**

Cuando tenemos múltiples versiones de Python instaladas podemos invocar una versión específica usando **py**. Por ejemplo, para invocar la versión 3.12 utilizamos el siguiente comando:

```bash
py -V:3.12
```

Si no se indica una versión específica entonces se ejecuta la versión más reciente de las instsladas.

**4. Creación de un entorno virtual con una versión específica de Python**

Siempre es recomendable el empleo de [entornos virtuales en Python](https://docs.python.org/3/tutorial/venv.html). Para crear un entorno virtual llamado ***CUIA*** que use la versión de Python 3.12, el comando a emplear es:

```bash
py -V:3.12 -m venv CUIA
```

Cuando activemos el entorno virtual, las versiones de *python* y *py* disponibles serán las indicadas en la creación del entorno.
